## Limpieza de ambos Datasets

```python
print(f"Cantidad de datos ausentes en películas: {movies.isnull().sum().sum()}")
print(f"Cantidad de datos duplicados en películas: {movies.duplicated().sum()}")
print(f"\nCantidad de datos ausentes en series de tv: {tv.isnull().sum().sum()}")
print(f"Cantidad de datos duplicados en series de tv: {tv.duplicated().sum()}")
```

No hay ningún duplicado en los datasets, por lo que obviamos el manejo de estos.

```python
print("Conteo de valores 'NaN' en películas:\n")
conteo_Nan_movies = movies.isnull().sum()
conteo_Nan_tv = tv.isnull().sum()

porcentaje_Nan_movies = (conteo_Nan_movies / len(movies)) * 100
porcentaje_Nan_tv = (conteo_Nan_tv / len(tv)) * 100

print(conteo_Nan_movies[conteo_Nan_movies > 0].sort_values(ascending=False))

print("\nConteo de valores 'NaN' en series:\n")
print(conteo_Nan_tv[conteo_Nan_tv > 0].sort_values(ascending=False))

print("\nPorcentaje NaN en películas (%):\n")
print(porcentaje_Nan_movies[porcentaje_Nan_movies > 0].sort_values(ascending=False).round(2))

print("\nPorcentaje NaN en series (%):\n")
print(porcentaje_Nan_tv[porcentaje_Nan_tv > 0].sort_values(ascending=False).round(2))

plt.figure(figsize=(10, 4))
sns.heatmap(movies.isnull(), cbar=False, yticklabels=False, cmap='viridis')

plt.title("Mapa de calor valores 'NaN' en películas")
plt.show()

plt.figure(figsize=(10, 4))
sns.heatmap(tv.isnull(), cbar=False, yticklabels=False, cmap='magma')

plt.title("Mapa de calor valores 'NaN' en series de tv")
plt.show()
```

## Manejo de Outliers

```python
cols_numericas_movies = movies.select_dtypes(include=['number']).columns.tolist()
cols_numericas_tv = tv.select_dtypes(include=['number']).columns.tolist()
```

```python
movies_sin_outliers = movies.copy()
tv_sin_outliers = tv.copy()

# Movies
for col in cols_numericas_movies:
    Q1 = movies_sin_outliers[col].quantile(0.25)
    Q3 = movies_sin_outliers[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    movies_sin_outliers[col] = np.clip(movies_sin_outliers[col], lower_bound, upper_bound)


fig, axs = plt.subplots(len(cols_numericas_movies), 2, figsize=(16, 5 * len(cols_numericas_movies)))
for i, col in enumerate(cols_numericas_movies):

    nombre_columna = col.capitalize()

    ax_hist_orig = axs[i, 0]
    ax_hist_orig.set_title(f'Distribución de {nombre_columna}\n(Original)', fontweight='bold', fontsize=13)

    sns.histplot(movies[col].dropna(), bins=30, kde=True, ax=ax_hist_orig, color='indianred', stat='density')
    ax_hist_orig.set_xlabel('')
    ax_hist_orig.tick_params(labelbottom=False)
    ax_box_orig = ax_hist_orig.inset_axes([0, -0.22, 1, 0.2])
    sns.boxplot(x=movies[col].dropna(), ax=ax_box_orig, color='indianred', fliersize=3)

    ax_box_orig.set_xlim(ax_hist_orig.get_xlim())
    ax_box_orig.set_xlabel(f'Valor de {nombre_columna}', fontweight='bold')

    ax_hist_cap = axs[i, 1]
    ax_hist_cap.set_title(f'Distribución de {nombre_columna}\n(Atípicos Recortados)', fontweight='bold', fontsize=13)

    sns.histplot(movies_sin_outliers[col].dropna(), bins=30, kde=True, ax=ax_hist_cap, color='mediumseagreen', stat='density')
    ax_hist_cap.set_xlabel('')
    ax_hist_cap.tick_params(labelbottom=False)

    ax_box_cap = ax_hist_cap.inset_axes([0, -0.22, 1, 0.2])
    sns.boxplot(x=movies_sin_outliers[col].dropna(), ax=ax_box_cap, color='mediumseagreen', fliersize=3)

    ax_box_cap.set_xlim(ax_hist_cap.get_xlim())
    ax_box_cap.set_xlabel(f'Valor de {nombre_columna} (Con Capping)', fontweight='bold')

plt.subplots_adjust(hspace=0.7)
plt.show()
```

```python
#TV series
for col in cols_numericas_tv:
    Q1 = tv_sin_outliers[col].quantile(0.25)
    Q3 = tv_sin_outliers[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    tv_sin_outliers[col] = np.clip(tv_sin_outliers[col], lower_bound, upper_bound)


fig, axs = plt.subplots(len(cols_numericas_tv), 2, figsize=(16, 5 * len(cols_numericas_tv)))
for i, col in enumerate(cols_numericas_tv):

    nombre_columna = col.capitalize()

    ax_hist_orig = axs[i, 0]
    ax_hist_orig.set_title(f'Distribución de {nombre_columna}\n(Original)', fontweight='bold', fontsize=13)

    sns.histplot(tv[col].dropna(), bins=30, kde=True, ax=ax_hist_orig, color='indianred', stat='density')
    ax_hist_orig.set_xlabel('')
    ax_hist_orig.tick_params(labelbottom=False)
    ax_box_orig = ax_hist_orig.inset_axes([0, -0.22, 1, 0.2])
    sns.boxplot(x=tv[col].dropna(), ax=ax_box_orig, color='indianred', fliersize=3)

    ax_box_orig.set_xlim(ax_hist_orig.get_xlim())
    ax_box_orig.set_xlabel(f'Valor de {nombre_columna}', fontweight='bold')

    ax_hist_cap = axs[i, 1]
    ax_hist_cap.set_title(f'Distribución de {nombre_columna}\n(Atípicos Recortados)', fontweight='bold', fontsize=13)

    sns.histplot(tv_sin_outliers[col].dropna(), bins=30, kde=True, ax=ax_hist_cap, color='mediumseagreen', stat='density')
    ax_hist_cap.set_xlabel('')
    ax_hist_cap.tick_params(labelbottom=False)

    ax_box_cap = ax_hist_cap.inset_axes([0, -0.22, 1, 0.2])
    sns.boxplot(x=tv_sin_outliers[col].dropna(), ax=ax_box_cap, color='mediumseagreen', fliersize=3)

    ax_box_cap.set_xlim(ax_hist_cap.get_xlim())
    ax_box_cap.set_xlabel(f'Valor de {nombre_columna} (Con Capping)', fontweight='bold')

plt.subplots_adjust(hspace=0.7)
plt.show()
```

## Implementación de reglas de negocios

### Filtrado financiero

Para que una película se considere rentable, su recaudación debe duplicar el presupuesto.

```python
movies_rentables = movies[(movies['budget'] > 0) & (movies['revenue'] > movies['budget']*2)].copy()
movies_rentables.count()
```

Para determinar las mejores series de tv, se busca que esta tenga más de 100 votos y que tengan un promedio mayor a 7.0

```python
tv_best = tv[(tv['vote_count'] > 100) & (tv['vote_average'] > 7.0)].copy()
```

### Normalización/Escalas

Se convierten en millones para mejor lectura en reportes

```python
movies_rentables['budget_millions'] = round(movies_rentables['budget'] / 1e6,2)
movies_rentables['revenue_millions'] = round(movies_rentables['revenue'] / 1e6,2)
```

### Cálculo de variables derivadas/KPIs

Se calcula el retorno de inversión (ROI)

```python
movies_rentables['ROI_pct'] = round(((movies_rentables['revenue'] - movies_rentables['budget']) / movies_rentables['budget']) * 100,3)
```

```python
movies_rentables.head()
```

Proporción de popularidad en relación con la cantidad de votos reales


```python
tv_best['engagement_ratio'] = np.where(
    tv_best['vote_count'] > 0,
    tv_best['popularity'] / tv_best['vote_count'],
    0
)
```

```python
#Retorno de inversión

df_limpio = movies[(movies['budget'] > 0) & (movies['revenue'] > 0)].copy()
df_limpio['ROI_pct'] = ((df_limpio['revenue'] - df_limpio['budget']) / df_limpio['budget']) * 100


df_rentables = df_limpio[df_limpio['ROI_pct'] > 0].copy()

limites = [0, 100, 300,500,1000, np.inf]
nombres_rangos = [
    '0% - 100%\n(Retorno Bajo)',
    '100% - 300%\n(Retorno Medio)',
    '300% - 500%\n(Alta Rentabilidad)',
    '500% - 1000%\n(Grandes Éxitos)',
    '+1000%\n(Súper Éxitos)'
]

df_rentables['Nivel_ROI'] = pd.cut(df_rentables['ROI_pct'], bins=limites, labels=nombres_rangos)

conteo_roi = df_rentables['Nivel_ROI'].value_counts().reset_index()
conteo_roi.columns = ['Rango de Rentabilidad', 'Cantidad de Títulos']

conteo_roi['Rango de Rentabilidad'] = pd.Categorical(
    conteo_roi['Rango de Rentabilidad'],
    categories=nombres_rangos,
    ordered=True
)
conteo_roi = conteo_roi.sort_values('Rango de Rentabilidad')

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x='Rango de Rentabilidad',
    y='Cantidad de Títulos',
    data=conteo_roi,
    palette='viridis'
)

plt.title('Volumen de Títulos por Nivel de Rentabilidad (ROI)', fontsize=14, weight='bold', pad=20)
plt.xlabel('Rango de Retorno de Inversión', fontsize=11, weight='bold', labelpad=10)
plt.ylabel('Cantidad de Películas', fontsize=11, weight='bold')

sns.despine()

for container in ax.containers:
    ax.bar_label(container, padding=5, fontsize=11, weight='bold', fmt='%d')

plt.tight_layout()
plt.show()
```
